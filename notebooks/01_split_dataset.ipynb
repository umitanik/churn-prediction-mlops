# Bank Customer Churn Prediction Model

We will develop a model to detect customer **churn** in the banking system.

### Project Scope:
* This model will be trained using only **90%** of the data source.
* The remaining **10%** will be reserved for testing purposes as a simulation of real-world data.

---
**Dataset Source:** [Kaggle - Bank Customer Churn](https://www.kaggle.com/datasets/radheshyamkollipara/bank-customer-churn/data)

> **Note:** this notebook documents how the split is produced. The canonical, runnable implementation lives in [`src/data/split_dataset.py`](../src/data/split_dataset.py) (`python -m src.data.split_dataset`) and is what the container bootstrap calls. Both use `test_size=0.1, random_state=42` and produce identical files.

```python
# Import necessary libraries
import os

import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# Load the dataset and check its shape
data = pd.read_csv('../data/raw/Customer-Churn-Records.csv')
data.shape
```

```python
# Display the first 5 rows of the dataset
data.head()
```

```python
# Split the dataset into training and testing sets with shuffling
train_data, test_data = train_test_split(data, test_size=0.1, random_state=42, shuffle=True)
```

```python
# Check the shapes of the training and testing sets
train_data.shape, test_data.shape
```

```python
# Save the training set to a CSV file
os.makedirs('../data/processed', exist_ok=True)
train_data.to_csv('../data/processed/train.csv', index=False)
```

```python
# Save the test set without the target variable, and save the ground truth separately
test_data.drop(columns="Exited").to_csv('../data/processed/test.csv', index=False)
test_data[["Exited"]].to_csv('../data/processed/ground_truth.csv', index=False)
```

```python
# Load the saved CSV files
train_data = pd.read_csv('../data/processed/train.csv')
test_data = pd.read_csv('../data/processed/test.csv')
ground_truth = pd.read_csv('../data/processed/ground_truth.csv')
```
